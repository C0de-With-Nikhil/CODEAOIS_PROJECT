# **1. Descriptive Statistics: Understanding Our Classroom**

**In this section, we are analyzing data we already have. We have a single classroom of 10 students, and we want to find the average score, the middle score, and how spread out the grades are.**

* **We import `numpy`, which is the industry-standard data science library for
handling math and arrays.**


* **We use `np.mean()` to calculate the Mean (Average) by adding all scores and dividing by 10.**


* **We use `np.median()` to find the Median (Middle Score) when the grades are sorted from lowest to highest.**


* **We use `np.std()` to find the Standard Deviation. This tells us the spread. A low number means everyone scored similarly; a high number means a massive gap between top and bottom students.**

```python
import numpy as np

# 1. Our raw data (10 students in our classroom)
classroom_scores = [85, 92, 78, 90, 88, 76, 95, 89, 82, 91]

# 2. Calculating the metrics
average_score = np.mean(classroom_scores)
middle_score = np.median(classroom_scores)

# 3. Calculating the spread of data
grade_spread = np.std(classroom_scores)

print(f"The Average (Mean) Score is: {average_score}")
print(f"The Median Score is: {middle_score}")
print(f"The Standard Deviation is: {grade_spread:.2f}")
```

#**Why Do We Need Standard Deviation?**


```python
import numpy as np

# Let's prove our Whiteboard theory!
class_A = [50, 50, 50, 50, 50]
class_B = [0, 0, 50, 100, 100]

print("--- CLASS A ---")
print(f"Average: {np.mean(class_A)}")
print(f"Standard Deviation (Spread): {np.std(class_A):.2f}")

print("\n--- CLASS B ---")
print(f"Average: {np.mean(class_B)}")
print(f"Standard Deviation (Spread): {np.std(class_B):.2f}")
```

# **2. Inferential Statistics: Predicting the Entire University**
**Now, we don't have the data for the entire university (5,000 students). We only have our 10 classroom students. We want to use our small sample to make an educated guess about the massive population.**

```python
import numpy as np

# 1. Creating a hidden population of 5000 university students (True average = 85)
university_population = np.random.normal(loc=85, scale=5, size=5000) #Mean (μ): 85  Standard Deviation (σ): 5 Total Sample Size: 5,000

# 2. Selecting a random sample of 10 students from that population
classroom_sample = np.random.choice(university_population, size=1000)

# 3. Calculate the sample average (Our Guess)
sample_mean = np.mean(classroom_sample)

# 4. Calculate the real population average (The Truth)
true_mean = np.mean(university_population)

print(f"Our 1000 students' average (Our Guess): {sample_mean:.2f}")
print(f"The actual 5000 students' average (The Truth): {true_mean:.2f}")
```

* **Since we can't interview 5,000 people in real life, we use `np.random.choice()` to randomly pick 10 students to represent our classroom sample.**



* **We calculate the average of just our 10 random students `(sample_mean)`. This is our "guess".**






* **We calculate the actual average of all 5,000 students `(true_mean)` to see if our inferential guess was accurate.**


# ** How to Make Our Guess More Accurate? **
**If a sample of 10 students gives us a rough guess, what happens if we look at a larger group? Let's increase our sample size to 100 students and see if our inference gets closer to the true population average of 85**.

```python
# Increasing the sample size to 100 students
better_sample = np.random.choice(university_population, size=100)
better_mean = np.mean(better_sample)

print(f"Our new guess with 100 students: {better_mean:.2f}")
print(f"Reminder - The True Population Average is: {true_mean:.2f}")
```
